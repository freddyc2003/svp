```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({'font.size': 12})
```

```python
# Load the CSV files
df_075 = pd.read_csv('../accuracy-results-0.75.csv', header=None)
df_099 = pd.read_csv('../accuracy-results-0.99.csv', header=None)

# Define the column names
column_names = ['Dimensions', 'Bit Level', 'Test Type', 'Lattice Basis', 
                'My Solution', 'Fplll Solution', 'Absolute Error']

# Assign the column names to the dataframes
df_075.columns = column_names
df_099.columns = column_names

# Add a new column for delta values
df_075['Delta'] = 0.75
df_099['Delta'] = 0.99

# Combine the two dataframes into one
combined_df = pd.concat([df_075, df_099])
```

```python
# Plotting the graph
plt.figure(figsize=(10, 6))
    
sns.lineplot(data=combined_df, x='Dimensions', y='Absolute Error', hue='Delta', palette="tab10")

# plt.title('Absolute Error vs Dimensions for Different Delta Values')
plt.xlabel('Dimensions')
plt.ylabel('Absolute Error')
plt.legend(title='Delta')

plot_filename = 'accuracy_absolute_error_vs_dimensions_plot.png'
plt.savefig(plot_filename, dpi=800)

plt.show()
```
